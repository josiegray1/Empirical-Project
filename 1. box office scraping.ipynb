# Scraping box office 


```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import time
import pandas as pd
#chrome driver path 
service=Service('/Users/jocelynegray/Library/Mobile Documents/com~apple~CloudDocs/Downloads/chromedriver-mac-x64/chromedriver')
driver=webdriver.Chrome(service=service)
#initialising an empty list to store the data
all_movies=[]
#entering the URL that im scraping 
base_url='https://www.the-numbers.com/box-office-records/worldwide/all-movies/cumulative/all-time/'
#loop through all pages
for page_num in range(0, 6001, 100):  #i chose to do 6000 films 
    url = base_url + str(page_num)
    print(f"Scraping page: {url}")
    driver.get(url)
    time.sleep(3)
    #extracting using beautiful soup 
    soup=BeautifulSoup(driver.page_source, 'html.parser')
    #find the table (info is stored in a table)
    table=soup.find('table')
    #getting all movie rows from the table but skipping the header 
    rows=table.find_all('tr')[1:]
    #looping through rows and extract film title and worldwide gross
    for row in rows:
        cols = row.find_all('td')
        if len(cols) > 1:  
            rank = cols[0].text.strip()  
            year = cols[1].text.strip()  
            title = cols[2].text.strip()  
            worldwide_gross = cols[3].text.strip()  
            all_movies.append({'rank': rank,'year': year,'title': title,'worldwide_gross': worldwide_gross})
#putting it into a dataframe 
movies_df = pd.DataFrame(all_movies)
# saving to a csv file 
movies_df.to_csv('box_office.csv', index=False)
print(movies_df.head(20))
#closing the browser 
driver.quit()

```

# cleaning and creating worldwide_box_office.csv

```python
df = pd.read_csv('box_office.csv')
#removing dollar signs so its numeric
df['worldwide_gross']=df['worldwide_gross'].replace('[\$,]', '',regex=True).astype(float)
df.to_csv('worldwide_box_office_movies.csv', index=False)
print(df.head())
```
